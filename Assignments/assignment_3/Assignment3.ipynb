(c) Plot a Normal Random Variable
<font size = "3">

To get full points:

- Set the seed using `np.random.seed(1234)`
- Make the random variable with the following parameters:
    -  mean = 7
    - standard deviation = 5
    - size = 1500

- Make a histogram
- Create a title
- Label the axes

```python
# Write your answer here
#set the seed
num.random.seed(1234)
mean = 7
standard_dev = 5
size = 1500

standard_normal_samples = num.random.normal(loc = mean, scale = standard_dev, size = size)
mat.hist(x=standard_normal_samples)
mat.xlabel("normal samples")
mat.ylabel("frequency")
mat.title("histogram of the normal samples")
mat.show()

```
